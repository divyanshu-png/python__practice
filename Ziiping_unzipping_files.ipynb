```python
import zipfile
comp_file= zipfile.ZipFile('comp_file.zip', 'w')
```

```python
#zipping file1 and file2 in the comp_file
comp_file.write('file1.txt', compress_type= zipfile.ZIP_DEFLATED)
comp_file.write('file2.txt', compress_type=zipfile.ZIP_DEFLATED)

comp_file.close()
```

```python
#extracting items from the zip filter
zip_obj= zipfile.ZipFile('comp_file.zip','r')
zip_obj.extractall("extracted_content")
```

```python
import re 
def search(file, pattern=r'\d{3}-\d{3}-\d{4}'):
    f= open(file, 'r')
    text= f.read()
    if re.search(pattern, text):
        return re.search(pattern, text)
    else:
        return ''
```

```python
results=[]
for folder, subfolders, files in os.walk(r'D:\New folder (3)\PYTHON\Unziped_for_instructions'):
    for f in files:
        full_path= folder+'\\'+f
        results.append(search(full_path))
```

```python
for r in results:
    if r !=' ':
        print(r)
    else: pass
```

```python
import re
import os 
for folder, subfolders, files in os.walk(r'D:\New folder (3)\PYTHON\Unziped_for_instructions'):
    print(f"Currently looking at {folder} \n")
    print('The subfolders are :')
    for sub_fold in subfolders:
        print(f"Subfolder: {sub_fold}")
    print('\n')
    for f in files:
        print(f"\t File: {f}")

        filename = f  # Define filename explicitly
        if os.path.exists(filename):
            with open(filename, 'r+') as f:
                # Perform read/write operations here
                lines= f.readline()
                re.findall(r'\d{3}-\d{3}-\d{4}', lines)
        else:
            print(f"File '{filename}' not found.")
        
        
    print('\n')
```
